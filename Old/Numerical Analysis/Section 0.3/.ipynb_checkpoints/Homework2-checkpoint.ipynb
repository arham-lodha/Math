```python
from scipy.optimize import fsolve
from sympy import *
import numpy as np
import math
from IPython.display import display, Markdown, Latex
```

### Section 1.3: Computer Problem 2
**Problem**: Let $f(x) = \sin(x^3) - x^3$. (a) Find the multiplicity of the root $r = 0$. (b) Use fzero method with initial guess $x = 0.1$ to locate the root. What are the forward and backward errors of fzero's response?

```python
def f(x):
    return np.sin(np.power(x, 3)) - np.power(x, 3)
```

```python
str_ = ""
multiplicity = 0

emach = 2e-53

x = Symbol('x')
g = sin(x**3) - x**3
val = g.evalf(subs={x:0})

while True:
    multiplicity += 1
    g = diff(g, x)
    val = g.evalf(subs={x:0})

    display(g)

    if not -emach <= val <= emach:
        display(Latex(f"$f^{{({multiplicity})}}(0) = {val}$"))
        break
    else:
        display(Latex(f"$f^{{({multiplicity})}}(0) = 0$"))


str_ += f"1. Multiplicity of f at the root r = 0 is {multiplicity}"

print(str_)
# markdown_text = markdown.markdown(str_)
```

```python
r = 0
```

```python
approx_root = fsolve(f, 0.1)[0]
```

```python
forward_error = abs(r-approx_root)
f"The forward error is f{forward_error}"
```

```python
backward_error = abs(f(approx_root))
backward_error
f"The backward error is f{backward_error}"
```

```python
display(Markdown(f"""
The approximate root is ${approx_root}$. The forward error is $\mid 0 - {approx_root}\mid = {forward_error}$. The backward error is $\mid f({approx_root}) \mid$ = {backward_error}.
"""))
```
