```julia
using Markdown
using Plots
```

```julia
function calculate_error(f_n_diff::Function, f::Function,  F::Function, x::Float64, h::Vector{Float64})
    return F.(f, x, h), f_n_diff(x) .- F.(f, x, h)
end
```

```julia
F(f::Function, x::Float64, h::Float64) = (f(x - h) - 2f(x) + f(x + h))/(h^2)
f_2_diff(x) = -cos(x)
```

```julia
h = [10.0^(-i) for i in 1.0:12.0]
approx, err = calculate_error(f_2_diff, cos, F, 0.0, h)
```

```julia
plot(h, err, xscale=:log10)
```

```julia
plot(h, abs.(err), xscale=:log10)
```

```julia
result = """# Section 5.1: Problem 5a \n
``f(x) = \\cos(x)``

Approximation of 2nd derivative

"""

for i in 1:length(h)
    result *= "`` h =10^{-$(i)}``: Approximation = $(approx[i]), Error = $(err[i]) \n\n"
end

result *= """Minimum error occurs at ``h = 10^{-4} = \\epsilon^{\\frac{\\log(10^{-4})}{\\log{\\epsilon}}} \\approx \\epsilon^{0.2555} \\approx \\epsilon^{1/4}``"""

Markdown.parse(result)
```
