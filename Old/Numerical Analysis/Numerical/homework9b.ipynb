```julia
function composite_simpsons(f::Function, a::Number, b::Number, panels::Int64)
    h = (b - a) / (panels)
    
    # Evaluate the function at evenly spaced points using a range
    y = f.(a:h:b)

    # Simpson's rule formula
    result = h / 3 * (y[1] + y[end] + 2 * sum(y[3:2:end-2]) + 4 * sum(y[2:2:end-1]))

    return result
end
```

# Section 5.2: Computer Problem 5a

$y = x^3$

$y' = 3x^2$

$f(x) = \sqrt{1 + y'^2} = \sqrt{1 + 9x^4}$

Approximate $\int_{0}^{1} f(x) \,dx$ with Composite Simpson's method and 32 panels

```julia
Markdown.parse("""The arc length is approximately $(composite_simpsons(x -> sqrt(1 + 9x^4), 0, 1, 32))""")
```
