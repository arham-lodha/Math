```julia
function composite_simpsons(f::Function, a::Number, b::Number, panels::Int64)
    h = (b - a) / (panels)
    
    # Evaluate the function at evenly spaced points using a range
    y = f.(a:h:b)

    # Simpson's rule formula
    result = h / 3 * (y[1] + y[end] + 2 * sum(y[3:2:end-2]) + 4 * sum(y[2:2:end-1]))

    return result
end
```
